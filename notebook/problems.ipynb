## Binomial distribution

In probability theory and statistics, the binomial distribution is the discrete probability distribution that gives only two possible results in an experiment; for example, when you flip a coin, it can be either head or tails.

```python
from scipy.stats import binom

# Define all the probability functions related to this distribution

def dbinom(x, size, prob = 0.5):
    """
    Calculates the point estimate of the binomial distribution
    """
    result = binom.pmf(k = x, n = size, p = prob, loc = 0)

    return result

def pbinom(q, size, prob = 0.5):
    """
    Calculates the cumulative of the binomial distribution
    """
    result = binom.cdf(k = q, n = size, p = prob, loc = 0)

    return result

def qbinom(p, size, prob = 0.5):
    """
    Calculates the quantile function from the binomial distribution
    """
    result = binom.ppf(q = p, n = size, p = prob, loc = 0)

    return result

def rbinom(n, size, prob = 0.5):
    """
    Generates random variables from the binomial distribution
    """
    result = binom.rvs(n = size, p = prob, size = n)

    return result
```

```python
# Binomial(10, 0.2) distribution, where n = 10 is the number of trials and p = 0.2 is the probability of success in a single trial
import numpy as np

np.random.seed(42)

print(f"Probability that a Binomial(10, 0.2) takes the value 2: {dbinom(2, size = 10, prob = 0.2)}")
print(f"Probability that a Binomial(10, 0.2) will take a value less than 2: {pbinom(2, size = 10, prob = 0.2)}")
print(f"Which value of a Binomial(10, 0.2) has a cumulative probability of 0.9? {qbinom(0.9, size = 10, prob = 0.2)}")
print(f"Generate 2000 random values from a Binomial(10, 0.2) distribution: {rbinom(2000, size = 10, prob = 0.2)}")
```

```python
import matplotlib.pyplot as plt

np.random.seed(42)

size = 10
prob = 0.2
data = rbinom(2000, size = 10, prob = 0.2)

pmf = dbinom(data, size = size, prob = prob)
cdf = pbinom(data, size = size, prob = prob)
ppf = qbinom(data, size = size, prob = prob)

fig, axis = plt.subplots(3, 1, figsize = (5, 7))

axis[0].stem(data, pmf, "r-")
axis[1].stem(data, cdf, "b-")
axis[2].stem(data, ppf, "g-")

# Adjust the layout
plt.tight_layout()

# Show the plot
plt.show()
```

Write the above functions but for both normal and uniform distributions.

```python
# Import libraries
from scipy.stats import norm
from scipy.stats import uniform
import numpy as np
import matplotlib.pyplot as plt

#Set randomness seed
np.random.seed(42)
```

# NORMAL DISTRIBUTION

```python
def dnorm(data, mean, std):
    # Calculates the probability density function of the normal distribution
    result = norm.pdf(x = data, loc = mean, scale = std)
    return result
def pnorm(data, mean, std):
    # Calculates the cumulative of the normal distribution
    result = norm.cdf(x = data, loc = mean, scale = std)
    return result
def qnorm(prob_data, mean, std):
    # Calculates the quantile function from the normal distribution
    result = norm.ppf(q = prob_data, loc = mean, scale = std)
    return result
def rnorm(n, mean, std):
    # Generates random variables from the normal distribution
    result = norm.rvs(loc = mean, scale = std, size = n)
    return result
```

```python
# Initial parameters
value=0
mean=1.5
std=2
n=10000
prob_data=pnorm(value, mean, std)
```

```python
# Print values
print(f"Probability that a Normal({mean}, {std}) distribution takes the value {value}: {dnorm(value, mean, std)}")
print(f"Probability that a Normal({mean}, {std}) distribution will take a value less than {value}: {pnorm(value, mean, std)}")
print(f"Which value of a Normal({mean}, {std}) distribution has a cumulative probability of {prob_data}? {qnorm(prob_data, mean, std)}")
print(f"Generate {n} random values from a Normal({mean}, {std}) distribution: {rnorm(n, mean, std)}")
```

```python
#Plot functions
data =np.sort(rnorm(n, mean, std))

pdf = dnorm(data, mean, std)
cdf = pnorm(data, mean, std)
ppf = qnorm(data, mean, std)

fig, axis = plt.subplots(3, 1, figsize = (7, 7))

axis[0].plot(data, pdf, "r-")
axis[1].plot(data, cdf, "b-")
axis[2].plot(data, ppf, "g-")

axis[0].set_title("Probability Density Function - PDF")
axis[1].set_title("Cumulative Distribution Fuction - CDF")
axis[2].set_title("Probability Point Fuction - PPF (Inverse CDF)")

axis[0].set_xlabel("Value")
axis[1].set_xlabel("Value")
axis[2].set_xlabel("Cumulative Probability")

axis[0].set_ylabel("Probability")
axis[1].set_ylabel("Cumulative Probability")
axis[2].set_ylabel("Value")

plt.tight_layout() # Adjust the layout
plt.show() # Show the plot
```

# UNIFORM DISTRIBUTION

```python
def duniform(data, loc, scale):
    # Calculates the probability density function of the uniform distribution
    result = uniform.pdf(x = data, loc = loc, scale = scale)
    return result
def puniform(data, loc, scale):
    # Calculates the cumulative of the uniform distribution
    result = uniform.cdf(x = data, loc = loc, scale = scale)
    return result
def quniform(prob_data, loc, scale):
    # Calculates the quantile function from the uniform distribution
    result = uniform.ppf(q = prob_data, loc = loc, scale = scale)
    return result
def runiform(n, loc, scale):
    # Generates random variables from the uniform distribution
    result = uniform.rvs(loc = loc, scale = scale, size = n)
    return result
```

```python
# Initial parameters
value=20
low_value = 19
high_value = 27
loc=low_value
scale=high_value-low_value
n=10000
prob_data=puniform(value, loc, scale)
```

```python
# Print values
print(f"Probability that a Uniform({low_value}, {high_value}) distribution takes the value {value}: {duniform(value, loc, scale)}")
print(f"Probability that a Uniform({low_value}, {high_value}) distribution will take a value less than {value}: {puniform(value, loc, scale)}")
print(f"Which value of a Uniform({low_value}, {high_value}) distribution has a cumulative probability of {prob_data}? {quniform(prob_data, loc, scale)}")
print(f"Generate {n} random values from a Uniform({low_value}, {high_value}) distribution: {runiform(n, loc, scale)}")
```
